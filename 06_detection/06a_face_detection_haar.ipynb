```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

```python
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
```

```python
img = cv2.imread('sherlock.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img)
```

```python
def detect_and_display_faces(face_cascade, img):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, flags=cv2.CASCADE_SCALE_IMAGE)

    print('Number of detected faces:', len(faces))

    result = np.copy(img)
    faces_img = []
    for (x, y, w, h) in faces: 
        cv2.rectangle(result, (x, y), (x+w, y+h), (0, 255, 0), 3)
        faces_img.append(img[y:y+h, x:x+w, :])

    plt.imshow(result)
    plt.show()

# Використання функції
detect_and_display_faces(face_cascade, img)
```

```python
img = cv2.imread('hats.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
detect_and_display_faces(face_cascade, img)
```
